```python
# Convert Date_of_Journey to datetime
df["Date_of_Journey"] = pd.to_datetime(
    df["Date_of_Journey"],
    format="%d/%m/%Y"
)

# Create new features
df["Journey_Day"] = df["Date_of_Journey"].dt.day
df["Journey_Month"] = df["Date_of_Journey"].dt.month

# Check the result
df[["Date_of_Journey", "Journey_Day", "Journey_Month"]].head()
```

```python
# Convert Departure Time to datetime
df["Dep_Time"] = pd.to_datetime(df["Dep_Time"])

# Extract Departure Hour and Minute
df["Departure_Hour"] = df["Dep_Time"].dt.hour
df["Departure_Minute"] = df["Dep_Time"].dt.minute

# Display the result
df[["Dep_Time", "Departure_Hour", "Departure_Minute"]].head()
```

```python
# Extract only the time part from Arrival_Time
df["Arrival_Time"] = df["Arrival_Time"].str.split(" ").str[0]

# Convert to datetime
df["Arrival_Time"] = pd.to_datetime(df["Arrival_Time"])

# Extract Arrival Hour and Minute
df["Arrival_Hour"] = df["Arrival_Time"].dt.hour
df["Arrival_Minute"] = df["Arrival_Time"].dt.minute

# Display the result
df[["Arrival_Time", "Arrival_Hour", "Arrival_Minute"]].head()
```

```python
# Display some duration values
df["Duration"].head(10)
```

```python
# Replace missing hours or minutes with 0
df["Duration"] = df["Duration"].str.replace("h", "h ")
df["Duration"] = df["Duration"].str.replace("m", "m ")

# Extract hours
df["Duration_Hours"] = (
    df["Duration"]
      .str.extract(r'(\d+)h')
      .fillna(0)
      .astype(int)
)

# Extract minutes
df["Duration_Minutes"] = (
    df["Duration"]
      .str.extract(r'(\d+)m')
      .fillna(0)
      .astype(int)
)

# Total duration in minutes
df["Duration_Total_Minutes"] = (
    df["Duration_Hours"] * 60
    + df["Duration_Minutes"]
)

# Display result
df[["Duration",
    "Duration_Hours",
    "Duration_Minutes",
    "Duration_Total_Minutes"]].head()
```

```python
print(df["Total_Stops"].unique())
```

```python
# Convert Total_Stops into numerical values

df["Total_Stops"] = df["Total_Stops"].replace({
    "non-stop": 0,
    "1 stop": 1,
    "2 stops": 2,
    "3 stops": 3,
    "4 stops": 4
})

# Check the result
df[["Total_Stops"]].head()
```

```python
# One-Hot Encoding for Airline
airline = pd.get_dummies(df["Airline"], prefix="Airline", dtype=int)

# Display first few rows
airline.head()
```

```python
source = pd.get_dummies(df["Source"], prefix="Source", dtype=int)

source.head()
```

```python
destination = pd.get_dummies(df["Destination"], prefix="Destination", dtype=int)

destination.head()
```

```python
df = pd.concat(
    [df, airline, source, destination],
    axis=1
)

print(df.shape)
```
